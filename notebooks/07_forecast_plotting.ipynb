## Import Libraries and Data Loading

```python
# Basic Packages
import numpy as np
import pandas as pd
from statsmodels.graphics.tsaplots import plot_acf
import matplotlib.pyplot as plt

#Plotting
import plotly.graph_objects as go
from plotting import plot_vf2, plot_vf_combined
```

# ----Kern County----

## Loading Data

```python
kern_lstm = pd.read_csv(
    "results/lstm/kern_lstm_plot_data.csv"
)

kern_sarimax = pd.read_csv(
    "results/sarimax/kern_sarimax_test_preds.csv"
)
```

## Converting to proper DateTime

```python
kern_lstm["train_dates"] = pd.to_datetime(
    kern_lstm["train_dates"]
)

kern_lstm["test_dates"] = pd.to_datetime(
    kern_lstm["test_dates"]
)
```

## Plotting

```python
fig_kern = plot_vf_combined(
    "Kern",
    train_dates=kern_lstm["train_dates"].dropna(),
    train_actuals=kern_lstm["train_actuals"].dropna(),

    train_predictions=kern_lstm["train_predictions"].dropna(),
    test_dates=kern_lstm["test_dates"].dropna(),

    test_actuals=kern_lstm["test_actuals"].dropna(),
    test_predictions=kern_lstm["test_predictions"].dropna(),

    sarimax_test_predictions=kern_sarimax["SARIMAX_Test_Pred"]
)

fig_kern.show()
```

# ----Fresno County----

## Loading Data

```python
fresno_lstm = pd.read_csv(
    "results/lstm/fresno_lstm_plot_data.csv"
)

fresno_sarimax = pd.read_csv(
    "results/sarimax/fresno_sarimax_test_preds.csv"
)
```

## Converting to proper DateTime

```python
fresno_lstm["train_dates"] = pd.to_datetime(
    fresno_lstm["train_dates"]
)

fresno_lstm["test_dates"] = pd.to_datetime(
    fresno_lstm["test_dates"]
)
```

## Plotting

```python
fig_fresno = plot_vf_combined(
    county="Fresno",

    train_dates=fresno_lstm["train_dates"].dropna(),
    train_actuals=fresno_lstm["train_actuals"].dropna(),
    train_predictions=fresno_lstm["train_predictions"].dropna(),

    test_dates=fresno_lstm["test_dates"].dropna(),
    test_actuals=fresno_lstm["test_actuals"].dropna(),
    test_predictions=fresno_lstm["test_predictions"].dropna(),

    sarimax_test_predictions=
        fresno_sarimax["SARIMAX_Test_Pred"]
)

fig_fresno.show()
```
